```python
# Run this once to permanently update the DataFrame in your dict with short column names

dfs['seafood_and_fish_production_thousand_tonnes'].rename(
    columns={
        "Commodity Balances - Livestock and Fish Primary Equivalent - Pelagic Fish - 2763 - Production - 5510 - tonnes": "Pelagic",
        "Commodity Balances - Livestock and Fish Primary Equivalent - Crustaceans - 2765 - Production - 5510 - tonnes": "Crustaceans",
        "Commodity Balances - Livestock and Fish Primary Equivalent - Cephalopods - 2766 - Production - 5510 - tonnes": "Cephalopods",
        "Commodity Balances - Livestock and Fish Primary Equivalent - Demersal Fish - 2762 - Production - 5510 - tonnes": "Demersal",
        "Commodity Balances - Livestock and Fish Primary Equivalent - Freshwater Fish - 2761 - Production - 5510 - tonnes": "Freshwater",
        "Commodity Balances - Livestock and Fish Primary Equivalent - Molluscs, Other - 2767 - Production - 5510 - tonnes": "Other_Molluscs",
        "Commodity Balances - Livestock and Fish Primary Equivalent - Marine Fish, Other - 2764 - Production - 5510 - tonnes": "Other_Marine"
    },
    inplace=True
)

dfs["fish_and_seafood_consumption_per_capita"].rename(columns={'Fish, Seafood- Food supply quantity (kg/capita/yr) (FAO, 2020)':'seafood_kg_per_capita_yr'},
inplace=True)
```

```python
df_prod = dfs['seafood_and_fish_production_thousand_tonnes']
df_cons = dfs['fish_and_seafood_consumption_per_capita']
df_main = dfs['capture_fisheries_vs_aquaculture']
```

```python
# Lowercase all column names + replace spaces with underscores
# Run this on each df (inplace)

def clean_column_names(df):
    df.columns = [col.lower().replace(' ', '_') for col in df.columns]
    return df

# Apply to all three
df_prod = clean_column_names(dfs['seafood_and_fish_production_thousand_tonnes'])
df_cons = clean_column_names(dfs['fish_and_seafood_consumption_per_capita'])
df_main = clean_column_names(dfs['capture_fisheries_vs_aquaculture'])

# Optional: also replace other common separators (commas, -, parentheses) if needed
def extra_clean(df):
    df.columns = [
        col.lower()
           .replace(' ', '_')
           .replace(',', '')
           .replace('-', '_')
           .replace('(', '')
           .replace(')', '')
           .replace('__', '_')   # avoid double underscores
        for col in df.columns
    ]
    return df

# If you want stricter cleaning:
df_prod = extra_clean(df_prod)
df_cons = extra_clean(df_cons)
df_main = extra_clean(df_main)
```

```python
# Reference the DataFrames
df_main = dfs['capture_fisheries_vs_aquaculture']
df_cons = dfs['fish_and_seafood_consumption_per_capita']
df_prod = dfs['seafood_and_fish_production_thousand_tonnes']

# Join the three tables
joined = (
    df_main
    .merge(df_cons, on=['entity', 'code', 'year'], how='inner')
    .merge(df_prod, on=['entity', 'code', 'year'], how='inner')
)

# Select the desired columns + create total production
final = joined[[
    'entity',
    'code',
    'year',
    'aquaculture_production_metric_tons',
    'capture_fisheries_production_metric_tons',
    'seafood_kg_per_capita_yr',
    'pelagic',
    'crustaceans',
    'cephalopods',
    'demersal',
    'freshwater',
    'other_molluscs',
    'other_marine'
]].copy()

# Add calculated total column
final['total_production_tonnes'] = (
    final['aquaculture_production_metric_tons'].fillna(0) +
    final['capture_fisheries_production_metric_tons'].fillna(0)
)

# Keep only rows with valid non-empty code
final = final[
    final['code'].notna() &
    (final['code'] != '') &
    (final['code'].str.strip() != '')
]

# Sort
final = final.sort_values(['entity', 'year']).reset_index(drop=True)

# Optional: reorder columns if you want total next to the two parts
final = final[[
    'entity', 'code', 'year',
    'aquaculture_production_metric_tons',
    'capture_fisheries_production_metric_tons',
    'total_production_tonnes',
    'seafood_kg_per_capita_yr',
    'pelagic', 'crustaceans', 'cephalopods', 'demersal',
    'freshwater', 'other_molluscs', 'other_marine'
]]

# Preview
final.head(15)

# Optional saves
# final.to_csv('fish_country_final_clean.csv', index=False)
# final.to_parquet('fish_country_final_clean.parquet')
```
